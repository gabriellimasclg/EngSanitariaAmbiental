```python
import pandas as pd
import numpy as np
```

```python
dados = pd.read_excel(r'C:\Users\gabriel.coimbra\Desktop\Meus arquivos\Tapes\Resultados Analíticos ETE Tapes (1).xlsx', skiprows=8, decimal = ',')
```

```python
dados = dados.dropna(axis=1, how='all')
dados.columns = dados.columns.str.replace('\n', ' ', regex=False)
dados = dados.drop(index=32)
```

```python
dados
```

```python
colunas_interesse = ['DBO5 mg O2/L', 'DQO mg O2/L', 'Ecoli NMP/100mL', 'NNH3 mg N/L','P tot mg P/L', 'pH',
                     'SST mg/L', 'T_Amb ºC', 'T_Am ºC', 'T_Receb °C','T_Receb ºC']

for coluna in colunas_interesse:
    dados[coluna] = dados[coluna].astype(str).str.replace('<', '', regex=False)
    dados[coluna] = dados[coluna].str.replace('>', '', regex=False)
    dados[coluna] = dados[coluna].str.replace('ND', '0', regex=False)
    dados[coluna] = dados[coluna].str.strip()
    dados[coluna] = pd.to_numeric(dados[coluna], errors='coerce')

media = dados[colunas_interesse].notna().astype(int).mean()
mediana = dados[colunas_interesse].notna().astype(int).median()
```
