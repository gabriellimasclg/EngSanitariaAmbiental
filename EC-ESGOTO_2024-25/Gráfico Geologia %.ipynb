```python
import matplotlib.pyplot as plt

# Dados
labels = [
    "Formação Rio do Rastro",
    "Gramado",
    "Formação Botucatu",
    "Formação Pirambóia",
    "Depósitos aluviais",
    "Depósitos Coluvios Aluviais"
]
sizes = [0.1615, 35.1578, 38.4072, 1.9047, 11.4943, 12.8744]
colors = ["#8EE9BA", "#DD8C76", "#EC82BB", "#D5D15D", "#A0DCEE", "#7FEC58", "#8A8DE9"]

# Criar gráfico de pizza
fig, ax = plt.subplots()
ax.pie(sizes, labels=[f"{l}\n{s:.1f}%" for l, s in zip(labels, [(s/sum(sizes))*100 for s in sizes])], 
       colors=colors, startangle=140, textprops={'color': 'white'},wedgeprops={'edgecolor': '#014B79'})

# Fundo transparente
fig.patch.set_alpha(0)
ax.set_facecolor('none')

# Mostrar gráfico
plt.savefig("geologia%",dpi=300, bbox_inches='tight')
plt.show()

```
