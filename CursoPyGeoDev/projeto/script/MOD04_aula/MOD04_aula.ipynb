# Vetores
Os vetores são arrays unidimensionais. Para criar um vetor atribuímos um nome a variável que vai receber um numpy array ou uma lista de valores.

vetor.png

```python
# Importe a biblioteca 
import numpy as np
```

```python
lista1 = [1, 2, 3, 4]

lista2 = np.array([1, 2, 3, 4]) #Python entende esse como um vetor
```

```python
print(lista1)
print()
print(lista2) #Python entende esse como um vetor
```

```python
print(type(lista1))
print()
print(type(lista2)) #Python entende esse como um vetor
```

```python
print(lista1 +3)
```

```python
print(lista2)
print(lista2 + 3) #conseguimos somar todos os valores por algo, ou outras operações
```

```python
print(lista2)
print(lista2 * 3)
```

```python
print(lista2)
print(lista2 / 3)
```

```python
print(lista2)
print(lista2 // 3)
```

```python
print(lista2.ndim) # um vetor tem apenas uma dimensão (por ex X)
```

# Operações de Matrizes
Uma matriz é semelhante a uma tabela com valores e se refere a um array com duas dimensões.
matriz.png

```python
# Criar matriz
matriz = np.array([lista1, lista1, lista1]) #duas dimensões, onde cada dimensão é uma "lista" #coloquei o np array na lista e transformou em array
matriz
print(matriz)
```

```python
matriz.shape
```

```python
matriz_zero = np.zeros((3, 3))
matriz_zero
```

```python
matriz_zero = np.zeros((3, 6))
matriz_zero
```

```python
# Soma
print(matriz)
print()
print(matriz + 3)
```

```python
# Subtração
print(matriz)
print()
print(matriz - 3)
```

```python
# Multiplicação
print(matriz)
print()
print(matriz * 2)
```

```python
# Divisão
print(matriz)
print()
print(matriz / 2)
```

```python
# Alterar valor na posição 1,1
print(matriz)
print()
matriz[1,1] = 5
print(matriz)
```

# Geometrias
- `Point` - É um objeto que representa um ponto no espaço.
- `LineString` -  é um objeto que representa uma sequência de pontos unidos para forma uma linha
- `Polygon` - É um objeto que representa uma área preenchida com pelo menos 3 tuplas de coordenadas formando o polígono.
- `Multipoint` - Representa uma coleção de pontos que representa uma lista de tuplas de coordenadas.
- `MultiLineString` - Representa uma coleção de linhas.
- `MultiPolygon` - Representa uma coleção de polígonos.
