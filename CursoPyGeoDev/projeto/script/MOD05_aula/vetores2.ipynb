```python
import geopandas as gpd
from tqdm import tqdm
import os

gdf = gpd.read_parquet(r"E:\arquivos_modulo_5\mpb_ro_2023.parquet")
gdf = gdf[gdf['DN'] == 39]
gdf.set_crs(epsg=4674,inplace=True,allow_override=True)

gdf2 = gpd.read_parquet(r"E:\arquivos_modulo_5\car_ro.parquet")
gdf.to_crs(epsg=4674)
print(gdf.crs)
#print(gdf2.crs)
```

```python
import os
from tqdm import tqdm

count = len(gdf2)
print(count)
num_linhas = tuple(range(0,count))

for i in tqdm(num_linhas):
    current_geometry = gpd.GeoDataFrame(geometry=[gdf2.geometry.iloc[i]], crs=gdf2.crs)

    joined = gpd.sjoin(gdf, current_geometry, how="inner", predicate="intersects")
    
    cod_imovel = gdf2['cod_imovel'].iloc[i]

    filename = f"{cod_imovel}_mpb.geojson"

    if not joined.empty:
        joined.to_file(os.path.join(r'E:\arquivos_modulo_5\car_x_mpb',filename), driver='GeoJSON')
```
