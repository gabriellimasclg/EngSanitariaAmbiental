```python
import pandas as pd
import pyproj

# Assuming 'resultado' is a pandas DataFrame with columns 'Latitude', 'Longitude', 'x', and 'y'
transformer = pyproj.Transformer.from_crs("epsg:4326", "epsg:32722", always_xy=True)

# Check if any values in the 'Latitude' column are NaN
if resultado['Latitude'].isna().any():
    # Apply the coordinate transformation and assign the results to 'Latitude' and 'Longitude'
    resultado['Longitude'], resultado['Latitude'] = zip(*resultado.apply(lambda row: transformer.transform(row['y'], row['x']), axis=1))

# Display the updated 'resultado' DataFrame
print(resultado)
```
