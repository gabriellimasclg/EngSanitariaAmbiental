```python
# método de fazer o geotiff para
# ainda não vejo aplicação

import os
from osgeo_utils.gdal_pansharpen import gdal_pansharpen

repopath = r'C:\Users\glima\OneDrive\Documentos\geral\CursoPyGeoDev\projeto'

pan_path = os.path.join(repopath,'inputs','CBERS_4A_WPM_20220902_205_133_L4_BAND0.tif')

rgb_path = os.path.join(repopath,'inputs','CBERS_4A_WPM_20220902_205_133_L4_RGB.tif')

output_path = os.path.join(repopath,'inputs','CBERS_4A_WPM_20220902_205_133_L4_RGB_PAN.tif')

args = [
    '', #tem q começar com a string vazia
    '-b', '1', #-b é banda, e o 1 é a banda 1
    '-b', '2',
    '-b', '3',
    '-threads', 'ALL_CPUS', # quantos núcleos do pc utilizar para o processamento
    pan_path, #banda monocromática
    rgb_path, #banda rgb
    output_path #saída
]

gdal_pansharpen(args)
print(f"Processo completo, arquivo salvo em {output_path}")
```
