```python
# Importa as bibliotecas necessárias
from osgeo import gdal
```

```python
# Chama a função de aplica o recorte no raster
input_raster = r'E:\CBERS_4A_WPM_20220902_205_133_L4_RGB_PAN.tif'

output_raster = r'E:\CBERS_4A_WPM_20220902_205_133_L4_RGB_PAN_mask.tif'

mask_shape = r'E:\mask.shp'

gdal.Warp(
    output_raster,
    input_raster,
    cutlineDSName=mask_shape,
    cropToCutline=True
)
```
