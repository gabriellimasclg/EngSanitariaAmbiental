```python
# Importa as bibliotecas a serem utilizadas
from osgeo import gdal
import os
import leafmap
import sys
import xarray
```

```python
# Aponta a localização dos arquivos
repopath = r'C:\Users\glima\OneDrive\Documentos\geral\CursoPyGeoDev\projeto\inputs'

band1_file = os.path.join(repopath,'CBERS_4A_WPM_20220902_205_133_L4_BAND1.tif')

band2_file = os.path.join(repopath,'CBERS_4A_WPM_20220902_205_133_L4_BAND2.tif')

band3_file = os.path.join(repopath,'CBERS_4A_WPM_20220902_205_133_L4_BAND3.tif')
```

```python
# Abre e lê as bnandas da imagem
band1_ds = gdal.Open(band1_file)

band2_ds = gdal.Open(band2_file)

band3_ds = gdal.Open(band3_file)
```

```python
# Armazena as infiormações da camada de entrada para a criação do raster
width = band1_ds.RasterXSize
height = band1_ds.RasterYSize
geotransform = band1_ds.GetGeoTransform()
projection = band1_ds.GetProjection()

print(geotransform)
```

```python
driver = gdal.GetDriverByName('Gtiff')
```

```python
# Cria o driver a aponta o local de saída do arquivo
output_file = r'C:\Users\glima\OneDrive\Documentos\geral\CursoPyGeoDev\projeto\outputs\CBERS_4A_WPM_20220902_205_133_L4_RGB.tif'
```

```python
# Cria o raster vazio para receber as 3 bandas
# gdal.GDT_Int16 --> isso é a resolução - profundidade do pixel
output_ds = driver.Create(output_file,width,height,3,gdal.GDT_Int16)
```

```python
# Aplica a matriz de transformação e configura o sistema de coordenadas
output_ds.SetGeoTransform(geotransform)
output_ds.SetProjection(projection)
```

```python
# Escreve cada banda de entrada no arquivo criado com 3 bandas
output_ds.GetRasterBand(1).WriteArray(band3_ds.ReadAsArray())
output_ds.GetRasterBand(2).WriteArray(band2_ds.ReadAsArray())
output_ds.GetRasterBand(3).WriteArray(band1_ds.ReadAsArray())
```

```python
# Adiciona um mapa com basemap
m = leafmap.Map(center=[-22,-51],zoom=18,height="800px")
m.add_basemap("SATELLITE")
m.to_html("map.html")
```

```python
m.layers[-1].visible = True
m.add_raster(output_file,band=[1, 2, 3], vmin=100, vmax=5000, layer_name="Image")
m.to_html(r'C:\Users\glima\OneDrive\Documentos\geral\CursoPyGeoDev\projeto\outputs\map.html')
```
