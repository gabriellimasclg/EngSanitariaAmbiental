```python
# Importa as bibliotecas necessárias
import leafmap
```

```python
# Aponta o caminhos dos arquivos
output_RGB_path = r'E:\CBERS_4A_WPM_20220902_205_133_L4_RGB.tif'

output_PAN_path = r'E:\CBERS_4A_WPM_20220902_205_133_L4_RGB_PAN_mask.tif'

output_NDVI_path = r'E:\CBERS_4A_WPM_20220902_205_133_L4_NDVI.tif'
```

```python
# Adiciona um mapa com basemap
m = leafmap.Map(center=[-22,-51],zoom=18,height="800px")
m.add_basemap("SATELLITE")
m
```

```python
# Adiciona um raster
m.layers[-1].visible = True
m.add_raster(output_RGB_path,band=[1, 2, 3], vmin=100, vmax=800, layer_name="RGB")

m.add_raster(output_PAN_path,band=[1, 2, 3], vmin=100, vmax=800, layer_name="PAN")
```

```python
# Adiciona um raster
m.layers[-1].visible = True
m.add_raster(output_NDVI_path,scheme='JenksCaspall',cmap='Greens',legend_title='NDVI')
```
